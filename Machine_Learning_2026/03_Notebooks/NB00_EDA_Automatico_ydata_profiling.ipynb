# EDA automatizado con `ydata-profiling`

## Exploración automática de un conjunto de datos

En este notebook utilizaremos **`ydata-profiling`** para generar automáticamente un informe exploratorio del conjunto de datos.

El objetivo es observar cómo una herramienta de EDA automatizado puede ayudarnos a detectar rápidamente características, patrones y posibles problemas de calidad en los datos.

> Un EDA automatizado **no reemplaza el análisis del estudiante o del analista**. La herramienta ayuda a detectar situaciones que posteriormente deben ser interpretadas, evaluadas y, cuando corresponda, tratadas mediante código.

## 2. Importar las librerías

```python
import pandas as pd
from ydata_profiling import ProfileReport
```

## 3. Cargar el dataset

Para este ejemplo utilizaremos directamente un archivo CSV publicado en GitHub.

No es necesario descargarlo previamente: `pandas` puede leerlo desde su URL.

```python
url = "https://raw.githubusercontent.com/juliopez/Taller-Fundamentos-Data-Science-Python/refs/heads/main/Machine_Learning_2026/02_Datasets/01_clientes_retail_video_limpieza.csv"

df = pd.read_csv(url)

df.head()
```

## 4. Exploración manual inicial

Antes de utilizar una herramienta automática, conviene realizar una inspección básica del conjunto de datos.

La idea es contrastar posteriormente lo que podemos observar manualmente con la información entregada por el perfilador automático.

```python
print("Dimensiones del dataset:", df.shape)
print("\nTipos de datos:")
display(df.dtypes)

print("\nPrimeros registros:")
display(df.head())

print("\nResumen estadístico:")
display(df.describe(include="all").T)
```

## 5. Generar el informe automático

`ProfileReport` analiza automáticamente el DataFrame y genera un informe que incluye, entre otros elementos:

- estructura general del dataset;
- tipos de variables;
- valores faltantes;
- valores únicos;
- distribuciones;
- estadísticas descriptivas;
- posibles duplicados;
- relaciones y correlaciones;
- advertencias o alertas sobre la calidad de los datos.

La generación del informe puede tardar algunos segundos.

```python
profile = ProfileReport(
    df,
    title="EDA automático — Clientes Retail",
    explorative=True
)

profile.to_notebook_iframe()
```

## 6. Exportar el informe a HTML

El informe también puede guardarse como un archivo HTML independiente para revisarlo posteriormente en un navegador.

```python
profile.to_file("EDA_clientes_retail.html")

print("Informe generado: EDA_clientes_retail.html")
```

# 7. ¿Qué debemos observar?

Revise el informe y trate de identificar situaciones que podrían requerir preparación o limpieza de datos.

Algunas preguntas útiles son:

- ¿Existen valores faltantes?
- ¿Aparecen registros duplicados?
- ¿Hay categorías escritas de manera inconsistente?
- ¿Existen valores poco plausibles o fuera de rango?
- ¿Qué variables son numéricas y cuáles categóricas?
- ¿Existen variables categóricas que deberían transformarse antes de utilizarlas en Machine Learning?
- ¿Las variables numéricas presentan escalas muy diferentes?

> Estas preguntas sirven para orientar la exploración. La presencia de una alerta en el informe **no significa automáticamente que exista un error** ni determina cuál debe ser la solución.

# 8. EDA automatizado vs. decisión analítica

El EDA automatizado permite acelerar la **detección** de posibles situaciones relevantes.

Sin embargo, todavía corresponde al analista:

1. comprobar que el problema existe;
2. interpretar su significado;
3. decidir si requiere tratamiento;
4. seleccionar una solución técnicamente adecuada;
5. implementar dicha solución;
6. verificar el resultado.

Podemos resumir el proceso como:

**Explorar → Detectar → Evidenciar → Decidir → Implementar → Justificar → Verificar**
